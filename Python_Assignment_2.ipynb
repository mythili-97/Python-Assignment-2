```python
#List Creation:
age_list=[24,25,27,28,29]
print("age_list:",age_list)
name_list=["Ram","Priya","Anu","Nivas","Prem"]
print("name_list:",name_list)
```

**Objective**:The objective of this assignment is to focus on lists,sets,dictionaries creation and modifications and conditional statements for decision-making.

Lists store multiple values in single variable.here list is created using both numbers and strings.

```python
#List Modifications:
name_list.append("Yazhini")
print(name_list)
age_list.insert(2,26)
print(age_list)
name_list.remove("Yazhini")
print(name_list)
age_list.pop()
print(age_list)
age_list.extend([30,31,32])
print(age_list)
age_list.sort(reverse=True)
print("Descending Order:", age_list)
```

Lists are modified using append,remove,pop,extend and sorted by descending order.

```python
print("Max age:",max(age_list))
print("Min age:",min(age_list))
print("Sum:",sum(age_list))
#Accessing list elements:
print("First element:",name_list[0])
print("Last element:",name_list[4])
print("Middle elements:",name_list[2:5])
name_list.reverse()
print("Reverse order:",name_list)
```

Elements in lists are accessed and maximum,minimum,average values are shown above.

```python
#Operators and conditional statements:
score=int(input("Enter your score"))
print(score)
if score>7:
  print("Great work,outstanding performance")
elif score<7 and score>4:
  print("Nice work,keep practicing")
elif score<4:
  print("Keep practicing,you can do better")
else:
  print("None of the above scores")


```
